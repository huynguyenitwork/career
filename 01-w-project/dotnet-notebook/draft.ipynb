```polyglot-notebook
using System.Text.Json;

public enum RK7Status
{
	NotFound = -1,
	Deleted = 0,
	Draft = 1,
	Inactive = 2,
	Active = 3
}
int? existingStatus;
var status = Enum.Parse<RK7Status>(string.IsNullOrEmpty(existingStatus?.ToString()) ? "-1" : existingStatus.ToString(), ignoreCase: true);
Console.WriteLine(status);
```
