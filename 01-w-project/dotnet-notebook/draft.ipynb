```polyglot-notebook
// mẫu kiểm tra convert type
public interface ISkyCrmDiscountCommandBase
{
    string Code { get; set; }
    string Name { get; set; }
}

public abstract class SkyCrmDiscountCommandBase : ISkyCrmDiscountCommandBase
{
    public string Code { get; set; }
    public string Name { get; set; }
    public string DiscountType { get; set; }
}
public class SkyCrmCreateDiscountCommand : SkyCrmDiscountCommandBase
{
    public string Description { get; set; }
}

SkyCrmDiscountCommandBase command = new SkyCrmCreateDiscountCommand{
    Code = "DISC10",
    Name = "10% Off",
    DiscountType = "Percentage",
    Description = "Get 10% off on all items."
};

if(command is SkyCrmCreateDiscountCommand baseCommand)
{
    Console.WriteLine($"Code: {baseCommand.Code}");
    Console.WriteLine($"Name: {baseCommand.Name}");
    Console.WriteLine($"Discount Type: {baseCommand.DiscountType}");
    Console.WriteLine($"Description: {baseCommand.Description}");
}
```
