## test int.parse vs int.tryparse

trường hợp cần test:
- null
- empty string
- giá trị hợp lệ
- giá trị sai format


```csharp
void TestTryParse(string input)
{
    var ok = int.TryParse(input, out int value);
    Console.WriteLine($"input: '{input ?? "null"}', ok: {ok}, value: {value}");
}

TestTryParse(null);
TestTryParse("");
TestTryParse("123");
TestTryParse("abc");
```

## test điều kiện if an toàn hơn

mục tiêu:
- tránh magic number
- đọc code là hiểu ý định


```csharp
int?[] cases = { null, 0, -1, 1, 10 };

foreach (var ident in cases)
{
    var isValid = ident is > 0;
    Console.WriteLine($"ident: {ident}, valid: {isValid}");
}
```

## mô phỏng logic giống production

giả lập hàm xử lý data từ api hoặc db


```csharp
string HandleIdent(string input)
{
    if (!int.TryParse(input, out var ident))
        return "invalid format";

    if (ident <= 0)
        return "ident not valid";

    return "ok";
}

var inputs = new[] { null, "", "0", "-5", "12", "abc" };

foreach (var i in inputs)
{
    Console.WriteLine($"{i ?? "null"} => {HandleIdent(i)}");
}
```
